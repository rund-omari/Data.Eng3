```python
import pandas as pd
data=pd.read_csv("https://raw.githubusercontent.com/rund-omari/Data.Eng3/refs/heads/main/datasets/jordan_retail_quiz_raw.csv")
```

```python
data_num=data.columns
data_num
```

```python
import matplotlib.pylab as plt
for i in data_num :
  amount=data[i].dropna()
  amount.hist(bins=20)
  plt.xlabel("Order Amount")
  plt.ylabel("Frecuency")
  plt.show()
  plt.boxplot(amount)
  plt.ylabel(i)
  plt.show()
```

```python
def detect(col):
  q1=col.quantile(.25)
  q3=col.quantile(.75)
  IRQ=q3-q1
  lower=q1-IRQ+1.5
  upper=q3+IRQ+1.5

  return lower,upper
```

```python
l , u = detect(data['order_amount_jod'])
```

```python
outlier = data[~(data['order_amount_jod'].between(l,u))]
outlier
```

```python
data["order_amount_jod"] = data["order_amount_jod"].clip(lower=1, upper=u)
```
