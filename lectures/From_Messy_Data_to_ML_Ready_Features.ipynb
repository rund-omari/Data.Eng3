```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
data = pd.read_csv('/content/jordan_retail_quiz_raw.csv')
```

```python
data.isna().sum()
```

```python
df_clean = data.copy()
```

```python
mean_amount = df_clean['order_amount_jod'].mean()
mean_amount
```

```python
meadian_amount = df_clean['order_amount_jod'].median()
meadian_amount
```

```python
#df_clean['order_amount_jod'] = df_clean['order_amount_jod'].fillna(meadian_amount)
df_clean.fillna({'order_amount_jod':meadian_amount},inplace=True)
```

```python
df_clean['order_amount_jod'].isna().sum()
```

```python
df_clean
```

```python
df_clean_2 = df_clean.dropna(subset=['order_date'])
df_clean_2
```

```python
df_clean_2['city'] = df_clean_2['city'].astype('string').str.strip().str.title()
df_clean_2['product_category'] = df_clean_2['product_category'].astype('string').str.strip().str.title()
```

```python
df_clean_2["order_date"] = pd.to_datetime(df_clean["order_date"],format="%Y-%m-%d",errors="coerce")
```
