```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler , StandardScaler
data=pd.read_csv("/content/SDK_Scaling_Visualization_Jordan_Ecommerce.csv")
```

```python
data.head()
```

```python
data.describe()
```

```python
data[['order_amount']]
```

```python
#from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(data[['order_amount']])
amount_scaling = scaler.transform(data[['order_amount']])
```

```python
amount = pd.DataFrame(amount_scaling)
amount.describe()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(4, 4))
plt.hist(data["order_amount"], bins=20, color="#E42BC5", alpha=0.5)
plt.xlabel("Order Amount (JOD)")
plt.ylabel("Frequency")
plt.title("Order Amount Distribution")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
sns.histplot(data["order_amount"], kde=True)
plt.show()
```

```python
import plotly.express as px
fig = px.histogram(data, x="order_amount", nbins=50,
                   title="Order Amount Distribution")
fig.update_layout(xaxis_title="Order Amount (JOD)", yaxis_title="Frequency")
fig.show()
```
